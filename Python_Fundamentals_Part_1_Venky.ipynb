# Branching

```python
# if-elif-else

a=7
b=6

if a>2:
    print("a is bigger than b")
elif a==2:
    print("a is equal to b")
elif a<2:
    print("a is less than b")
else:                                 #deadcode
    print("deadcode")         #deadcode
```

```python
# ternary
a=5
b=3
print('A') if a>b else print("B")   # here no indentiation needed. its recognized as sinle ternary operator statement
```

```python
# switch/case equivalent
day = 2
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
```

# Loops

```python
# For

for i in range(5):
    print(i,end=" ")
print()
for i in range(10,15):
    print(i,end=" ")
print()
for i in range(10,15,2):
    print(i,end=" ")
print()
for i in [2,4,5]:
    print(i,end=" ")
print()
for i in (2,4,5):
    print(i,end=" ")
print()
for i in "Hello":
    print(i,end=" ")

# continue is to skip to next iteration
# break is to break out of current loop
```

```python
c=1
while c<=5:
    c+=1
    if c==2:
        continue
    if c==4:
        break
    print(c,end=" ")
```
