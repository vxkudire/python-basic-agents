```python
import os
import json
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get('OPENAI_APIKEY'))
for model in client.models.list():
    print(model.id)
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Hello!"}
    ])
print(response.choices[0].message.content)
```

# Anthropic

```python
!pip install anthropic
```

```python
import os
import anthropic
from google.colab import userdata

client = anthropic.Anthropic(api_key=userdata.get('ANTHROPIC_APIKEY'))
for model in client.models.list():
    print(model.id)
```

```python
response = client.messages.create(
    model="claude-sonnet-4-6",
    max_tokens=200,
    messages=[
        {"role": "user", "content": "Hello Claude!"}
    ]
)

print(response.content[0].text)
```

# Gemini

```python
!pip install google-genai
```

```python
import os
from google import genai
from google.colab import userdata

os.environ["GEMINI_API_KEY"] = userdata.get('GEMINI_APIKEY')
client = genai.Client()
for model in client.models.list():
    print(model.name)
```

```python
response = client.interactions.create(
    model="gemini-3.6-flash",
    input="Explain quantum computing in one sentence."
)

print(response.output_text)
```

# Langchain

```python
!pip install langchain langchain-openai python-dotenv
```

```python
import os
from google.colab import userdata
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
    api_key=userdata.get('OPENAI_APIKEY')
)
```

```python
response = llm.invoke("Explain LangChain in one sentence.")

print(response.content)
```

## Using a Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} like I'm 10 years old."
)

chain = prompt | llm

response = chain.invoke({
    "topic": "Machine Learning"
})

print(response.content)
```

```python
prompt = ChatPromptTemplate.from_template(
"""
You are a {profession}.

Explain {topic}.
"""
)

chain = prompt | llm

response = chain.invoke({
    "profession": "teacher",
    "topic": "Neural Networks"
})

print(response.content)
```
