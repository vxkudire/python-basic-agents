### LeetCode 217

Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

```python
# Solution 1
def containsDuplicate(nums):
#     my_set = set(nums)
#     if len(my_set) == len(nums):
#         return False
#     return True
    return len(nums) != len(set(nums))
```

```python
# Solution 2
def containsDuplicate(nums):
    check_set = set()
    for num in nums:
        if num in check_set:
            return True
        else:
            check_set.add(num)
    return False
```

```python
# Test Case 1
nums = [1, 2, 3, 1]
print(containsDuplicate(nums))
```

```python
# Test Case 2
nums = [1, 2, 3, 4]
print(containsDuplicate(nums))
```

```python
# Test Case 3
nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]
print(containsDuplicate(nums))
```

### LeetCode 349

Given two integer arrays nums1 and nums2, return an array of their intersection. Each element in the result must be unique and you may return the result in any order.

```python
# Solution 1
def intersection(nums1, nums2):
    set1 = set(nums1)
    set2 = set(nums2)
    return list(set1.intersection(set2))
```

```python
# Solution 2
def intersection(nums1, nums2):
    return list(set(nums1) & set(nums2))
```

```python
# Test Case 1
nums1 = [1, 2, 2, 1] # {1, 2}
nums2 = [2, 2] # {2}
print(intersection(nums1, nums2))
```

```python
# Test Case 2
nums1 = [4, 9, 5] # {4, 9, 5}
nums2 = [9, 4, 9, 8, 4] # {9, 4, 8}
print(intersection(nums1, nums2))
```

#### Question in Class: How to import a user-defined module in python locally from another directory?

You can modify the Python path at runtime to include the directory where the module is located.

1. First, structure your project like this:

```python
# /project
# ├── /dir1
# │   └── my_module.py
# └── /dir2
#     └── script.py
```

2. In your script.py, you need to modify the sys.path to include the path to /dir1:

```python
import sys
import os

# Add the path of the directory where my_module.py is located
sys.path.append(os.path.abspath('../dir1'))

# Now you can import the module
import my_module
```

#### Question in Class: If I import numpy and pandas inside my custom-defined module and then import that module in my code, do I need to import numpy and pandas again?

No, you do not need to import NumPy or Pandas again in your main code if you are only using them inside your custom module.

**When you do need to import NumPy/Pandas in your main script:**

- If you call NumPy or Pandas functions directly in your main script.

- If you pass around or manipulate NumPy arrays or Pandas DataFrames using their functions.

#### How does importing modules work in venv?

Everything works as long as the venv is activated, because Python uses the venv’s version of numpy and pandas.

**What happens if you don't activate the venv?**

If you try to run the script with your system Python instead of the `venv`'s Python:

- Python won’t find the modules (unless they're also installed globally)

- You'll get `ModuleNotFoundError`

#### map()

The map() function in Python does not modify the original iterable in-place. Instead, it applies a specified function to each element of the iterable and returns an iterable containing the results. The original iterable remains unchanged.

```python
# Example - with normal Python function
numbers = [2, 4, 8, 6, 8, 10]

def square(number):
    return number ** 2

numbersSquare_list = list(map(square, numbers))
# numbersSquare_tuple = tuple(map(square, numbers))
# numbersSquare_set = set(map(square, numbers))

print(numbersSquare_list)
```

```python
# with Lambda function
numbersSquare_list = list(map(lambda number: number**2, numbers))
print(numbersSquare_list)
```

### LeetCode 169

```python
# Solution 1
def majorityElement(nums):
    count = 0
    candidate = None
    for i in range(len(nums)):
        if count == 0:
            candidate = nums[i]
        if candidate == nums[i]:
            count += 1
        else:
            count -= 1
    return candidate
```

```python
# Solution 2
def majorityElement(nums):
    freq = {}
    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
        if freq[num] > len(nums) // 2:
            return num
```

```python
# Test Case 1
nums = [3, 2, 3]
print(majorityElement(nums))
```

```python
# Test Case 2
nums = [2, 2, 1, 1, 1, 2, 2]
print(majorityElement(nums))
```

```python
# Test Case 2
nums = [2]
print(majorityElement(nums))
```

### Global vs Local

#### What is the output of this code?

```python
# What is the output?
# answer format: abcdefg

a = 1 # global variable

def f():
    print(a, end="")

def g():
    a = 2 # local variable
    print(a, end="")

def h():
    global a
    a = 3 # global variable
    print(a, end="")

print(a, end="")
f()
print(a, end="")
g()
print(a, end="")
h()
print(a)
```

#### What is the output of this code?

```python
# Format = a b c d e

x = 10 # global variable

def outer_function():

    y = 20 # local to outer_function

    def inner_function():
        global x
        nonlocal y

        x = 30 # global variable
        y = 40 # local to inner_function

        print(x, end=" ")
        print(y, end=" ")

    inner_function()
    print(x, end=" ")
    print(y, end=" ")

outer_function()
print(x)

# 30 40 30 20 30
```

### LeetCode 1

```python
# Solution 1
def twoSum(nums, target):
    dict1 = {}
    for i in range(len(nums)):
        dict1[nums[i]] = i
    for i in range(len(nums)):
        complement = target - nums[i]
        if complement in dict1 and dict1[complement] != i:
            return [i, dict1[complement]]
```

```python
# Solution 2
def twoSum(nums, target):
    dict1 = {}
    for i in range(len(nums)):
        complement = target - nums[i]
        if complement in dict1:
            return [i, dict1[complement]]
        dict1[nums[i]] = i
```

```python
# Test Case 1
nums = [2, 7, 11, 15]
target = 9
print(twoSum(nums, target))
```

```python
# Test Case 2
nums = [3, 2, 4]
target = 6
print(twoSum(nums, target))
```

```python
# Test Case 3
nums = [3, 3]
target = 6
print(twoSum(nums, target))
```
