##### <span style="background-color: yellow;">Qestion in Class: Is there any difference between '''...''' and """..."""?</span>

In Python, there is **no functional difference** between using single quotes (`'''...'''`) and double quotes (`"""..."""`) for multi-line strings. Both are used to define string literals that span multiple lines. However, there are some minor considerations:

1. **Multi-line Strings**:
   - Both `'''...'''` and `"""..."""` can be used to create strings that span multiple lines.

```python
 multi_line_str1 = '''This is a
 multi-line string.'''

 multi_line_str2 = """This is also a
 multi-line string."""
```

2. **Docstrings**:
   - Triple quotes (`'''` or `"""`) are typically used for docstrings in Python, which are used to document functions, classes, and modules.
   - The convention is to use `"""` for docstrings, although `'''` is also valid.

```python
def example_function():
    """
    This is a docstring.
    It describes the function's purpose.
    """
    pass
```

3. **Consistency**:
   - It is common practice to use double quotes (`"""`) for docstrings to maintain consistency with the Python community's style guide (PEP 257).
   - In contrast, for general multi-line strings, both single and double triple quotes are interchangeable, so you can use either depending on the context and preference.

4. **Escape Characters**:
   - The choice between `'''` and `"""` may also depend on the contents of the string. For example, if the string contains a single quote (`'`), it might be more convenient to use double quotes, and vice versa, to avoid the need for escaping quotes.

```python
string1 = """He said, "Python is awesome!" """
string2 = 'It\'s a beautiful day.'
```

##### Difference between (a == b) and (a is b)

In Python, the `id()` function returns the identity of an object, which is a unique identifier for that object during its lifetime in memory. This identity is typically represented as the memory address of the object.

```python
a = 2
b = 2

print(id(a))
print(id(b))
```

```python
print(a == b)
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(id(a))
print(id(b))
```

```python
print(a == b)
print(a is b)
```

**Conclusion:** Even though `a` and `b` contain the same values, they are **two separate objects** in memory. Therefore, their `id()` values will be different. This demonstrates that in Python, lists (and other mutable types) are objects that are stored in different memory locations even if their contents are identical.

##### <span style="background-color: yellow;">Question in Class: Main difference between copy and deepcopy?</span>

## Shallow Copy `copy.copy()`

- Creates a new object, but does not create copies of nested objects.

- It copies references to nested objects, not the actual objects.

- If you modify a nested object in the copy, it also affects the original.

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0][0] = 99
print(original)
```

## Deep Copy `copy.deepcopy()`

- Creates a new object, and recursively copies all nested objects.

- The result is fully independent of the original.

```python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0][0] = 99
print(original)
```

### Rule of Thumb

- Use `copy()` when you trust the internal structure or want to preserve shared references (**reuse**)

- Use `deepcopy()` only when you’re sure you need full independence and the object isn’t too large or complex (**protect**)

### LeetCode - Two Sum

Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

```python
# Solution 1
def twoSum(nums, target):
    n = len(nums)
    for i in range(n):
        num1 = nums[i]
        for j in range(i+1, n):
            num2 = nums[j]
            if num1 + num2 == target:
                return [i, j]
```

```python
# Solution 2
def twoSum(nums, target):
    for num in nums:
        diff = target - num
        if (diff in nums) and (diff != num):
            return [nums.index(num), nums.index(diff)]
```

```python
# Test Case 1
list1 = [2, 7, 11, 15]
result = twoSum(list1, 9)
print(result)
```

```python
# Test Case 2
list2 = [3, 2, 4]
result = twoSum(list2, 6)
print(result)
```

```python
# Test Case 3
list3 = [2, 10, 9, 20, 5, 7]
result = twoSum(list3, 11)
print(result)
```

### LeetCode - Contains Duplicate

Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

```python
# Solution 1
def containsDuplicate(nums):
    n = len(nums)
    for i in range(n-1):
        for j in range(i+1, n):
            if nums[i] == nums[j]:
                return True
    return False
```

```python
# Solution 2
def containsDuplicate(nums):
    nums.sort()
    n = len(nums)
    for i in range(n-1):
        if nums[i] == nums[i+1]:
            return True
    return False
```

```python
# Test Case 1
list1 = [1, 2, 3, 1]
result = containsDuplicate(list1)
print(result)
```

```python
# Test Case 2
list2 = [1, 2, 3, 4]
result = containsDuplicate(list2)
print(result)
```

```python
# Test Case 3
list3 = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]
result = containsDuplicate(list3)
print(result)
```

```python
# Test Case 4
list4 = [2, 10, 9, 20, 1, 7]
result = containsDuplicate(list4)
print(result)
```

### LeetCode - Revese Words in a String

Given an input string s, reverse the order of the words.

A word is defined as a sequence of non-space characters. The words in s will be separated by at least on space.

Return a string of the words in reverse order concatenated by a single space.

Note that s may contain leading or trailing spaces or multiple spaces between two words. The returned string should only have a single space separating the words. Do not include any extra spaces.

```python
def reverseWords(s):
    list_of_words = s.split()
    list_of_words.reverse()
#     list_of_words = list_of_words[::-1]
    new_str = " ".join(list_of_words)
    return new_str
```

```python
# Test Case 2
text2 = "  hello world  "
print(reverseWords(text2))
```

### LeetCode - Find Pivot Index

Given an array of integers nums, calculate the pivot index of this array.

The pivot index is the index where the sum of all the numbers strictly to the left of the index is equal to the sum of all the numbers strictly to the index's right.

If the index is on the left edge of the array, then the left sum is 0 because there are no elements to the left. This also applies to the right edge of the array.

Return the leftmost pivot index. If no such index exists, return -1.

```python
# Solution 1
def pivotIndex(nums):
    for i in range(len(nums)):
        l_sum = sum(nums[:i])
        r_sum = sum(nums[i+1:])
        if l_sum == r_sum:
            return i
    return -1
```

```python
# Solution 2
def pivotIndex(nums):
    S = sum(nums)
    left_sum = 0
    for i in range(len(nums)):
        right_sum = S - left_sum - nums[i]
        if left_sum == right_sum:
            return i
        left_sum += nums[i]
    return -1
```

```python
# Test Case 1
nums1 = [1, 7, 3, 6, 5, 6]
print(pivotIndex(nums1))
```
